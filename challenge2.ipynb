```python
backend = service.least_busy(
    operational=True,
    simulator=False,
    min_num_qubits=1
)

print("Using backend:", backend.name)
```

```python
from qiskit import QuantumCircuit

qc_0 = QuantumCircuit(1, 1)

qc_0.h(0)
qc_0.rz(1.0, 0)
qc_0.h(0)

qc_0.measure(0, 0)

print(qc_0)
```

```python
qc_1 = QuantumCircuit(1, 1)

qc_1.h(0)
qc_1.rz(2.14, 0)
qc_1.h(0)

qc_1.measure(0, 0)

print(qc_1)
```

```python
from qiskit import transpile

qc_0_transpiled = transpile(qc_0, backend)
qc_1_transpiled = transpile(qc_1, backend)

print("Both circuits transpiled successfully!")
```

```python
from qiskit_ibm_runtime import SamplerV2

sampler = SamplerV2(mode=backend)

job = sampler.run(
    [qc_0_transpiled, qc_1_transpiled],
    shots=100
)

print("Job submitted!")
print("Job ID:", job.job_id())
```

```python
result = job.result()

print("Job completed!")
```

```python
counts_0 = result[0].data.c.get_counts()
counts_1 = result[1].data.c.get_counts()

print("Circuit 1 results:", counts_0)
print("Circuit 2 results:", counts_1)
```

```python
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

plot_histogram(
    [counts_0, counts_1],
    legend=[
        "Circuit 1 - 0 biased",
        "Circuit 2 - 1 biased"
    ]
)

plt.show()
```
